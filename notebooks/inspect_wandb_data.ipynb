# Inspect WandB Data
This notebook helps inspect the neuron activation data from WandB runs

```python
import wandb
import pandas as pd

api = wandb.Api()
entity = "sparsey-testing-system"
```

```python
# Configure project and run
project_name = "post_presentation_4"
run_id = "d186v0fa"

run = api.run(f"{entity}/{project_name}/{run_id}")
```

```python
# Show all run history data
print("=== Complete Run History ===")
history = run.scan_history()
history_df = pd.DataFrame(history)

# Display all columns
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)

print("\nColumns in history:")
print(history_df.columns.tolist())

print("\nFull history data:")
print(history_df)

# Save to CSV for easier inspection
history_df.to_csv('wandb_history.csv')
print("\nSaved complete history to 'wandb_history.csv'")
```

```python
# Show summary data
print("=== Summary Data ===")
for key, value in run.summary.items():
    print(f"{key}: {value}")
```

```python
# Show config data
print("=== Config Data ===")
for key, value in run.config.items():
    print(f"{key}: {value}")
```

```python
# Get only raw activation columns
history = run.scan_history(keys=["num_activations/layer_1/mac_1",
                                "num_activations/layer_2/mac_2"])
df = pd.DataFrame(history)

print("\nRaw neuron activations:")
print(df)
```

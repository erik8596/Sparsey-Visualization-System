# Inspect WandB Data
This notebook helps inspect the neuron activation data from WandB runs

```python
import wandb
import pandas as pd
import matplotlib.pyplot as plt
import re

api = wandb.Api()
entity = "sparsey-testing-system"
```

```python
# Configure project and run
project_name = "post_presentation_4"
run_id = "e7pryimi"

run = api.run(f"{entity}/{project_name}/{run_id}")
```

```python
print("=== Neuron Activations by Step ===")
history = run.scan_history()

for step in history:
    neuron_activations = {key: value for key, value in step.items() if key.startswith('neuron_activations') and not key == 'neuron_activations'}
    if neuron_activations:
        print(f"Step {step['_step']}")
        print('-' * 28)
        for key in sorted(neuron_activations.keys()):
            if neuron_activations[key] == 1.0:
                print(key)
        print()
```

```python
print("=== Layer 2 MAC 1 CM 2 Neuron Activations by Step ===")
history = run.scan_history()

for step in history:
    neuron_activations = {key: value for key, value in step.items()
                        if key.startswith('neuron_activations/layer_2/mac_1/cm_2')}
    if neuron_activations:
        print(f"Step {step['_step']}")
        print('-' * 28)
        for key in sorted(neuron_activations.keys()):
            if neuron_activations[key] == 1.0:
                # Extract just the neuron number from the key
                neuron = key.split('/')[-1]
                print(f"{neuron}")
        print()
```

```python
print("=== Layer 2 MAC 1 Neuron Activations by Step ===")
history = run.scan_history()

for step in history:
    neuron_activations = {key: value for key, value in step.items()
                        if key.startswith('neuron_activations/layer_2/mac_4/')}
    if neuron_activations:
        print(f"Step {step['_step']}")
        print('-' * 28)
        # Group by CM
        cm_activations = {}
        for key in sorted(neuron_activations.keys()):
            if neuron_activations[key] == 1.0:
                # Extract CM and neuron number
                parts = key.split('/')
                cm = parts[-2]  # Get CM part (cm_X)
                neuron = parts[-1]  # Get neuron part (neuron_X)
                if cm not in cm_activations:
                    cm_activations[cm] = []
                cm_activations[cm].append(neuron)

        # Print activations grouped by CM
        for cm in sorted(cm_activations.keys()):
            print(f"{cm}:")
            for neuron in sorted(cm_activations[cm]):
                print(f"  {neuron}")
        print()
```

```python
print("=== All Data by Step ===")
history = run.scan_history()

for step in history:
    print(f"\nStep {step['_step']}")
    print('=' * 50)
    # Sort keys for better readability
    for key in sorted(step.keys()):
        if key != '_step':  # Skip the step number since we already printed it
            print(f"{key}: {step[key]}")
```

```python
print("=== Input Activations by Step ===")
history = run.scan_history()

for step in history:
    input_activations = {key: value for key, value in step.items()
                        if key.startswith('input_activations/')}
    if input_activations:
        print(f"\nStep {step['_step']}")
        print('-' * 28)
        for key in sorted(input_activations.keys()):
            if input_activations[key] == 1.0:
                # Extract just the input number from the key
                input_num = key.split('/')[-1]
                print(f"{input_num}")
```

# Inspect WandB Data
This notebook helps inspect the neuron activation data from WandB runs

```python
import wandb
import pandas as pd
import matplotlib.pyplot as plt
import re

api = wandb.Api()
entity = "sparsey-testing-system"
```

```python
# Configure project and run
project_name = "weights_testing"
run_id = "f3abeelg"

run = api.run(f"{entity}/{project_name}/{run_id}")
```

```python
# Show all run history data
print("=== Complete Run History ===")
history = run.scan_history()
history_df = pd.DataFrame(history)

# Display all columns
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)

print("\nColumns in history:")
print(history_df.columns.tolist())

print("\nFull history data:")
print(history_df)

# Save to CSV for easier inspection
history_df.to_csv('wandb_history.csv')
print("\nSaved complete history to 'wandb_history.csv'")
```

```python
print("=== Neuron Activations Through Each Step ===")
for step in history:
    activations = {key: value for key, value in step.items() if key.startswith('neuron_activations')}
    step_number = step.get('_step', 'Unknown')  # Use '_step' or fallback to 'Unknown'
    print(f"Step: {step_number}")
    for key, value in activations.items():
        print(f"  {key}: {value}")
    print("-" * 50)
```

```python
print("=== MAC Activations Through Each Step ===")
for step in history:
    activations = {key: value for key, value in step.items() if key.startswith('neuron_activations')}
    mac_activations = {key: value for key, value in activations.items()
                       if re.match(r"^neuron_activations/layer_\d+/mac_\d+$", key) and value == 1}
    step_number = step.get('_step', 'Unknown')  # Use '_step' or fallback to 'Unknown'
    print(f"Step: {step_number}")
    for key, value in mac_activations.items():
        print(f"  {key}: {value}")
    print("-" * 50)
```

```python
# print neuron activations with step number
print("=== Neuron Activations in Readable Format ===")
for step in history:
    activations = {key: value for key, value in step.items() if key.startswith('neuron_activations')}
    step_number = step.get('_step', 'Unknown')  # Use '_step' or fallback to 'Unknown'
    print(f"Step: {step_number}")
    for key, value in activations.items():
        if re.match(r"^neuron_activations/layer_\d+/mac_\d+/cm_\d+/neuron_\d+$", key):
            print(f"  {key}: {value}")
    print("-" * 50)
```

```python
print("=== Complete Run History ===")
history = run.scan_history()

print("\nDisplaying history step by step:")
for step in history:
    # Check for neuron activations in layer 2
    if 'neuron_activations/layer_2' in step:
        print(f"Step: {step['step']}, Neuron Activations (Layer 2): {step['neuron_activations/layer_2']}")

    print("-" * 50)
```

```python
# Extract and analyze neuron activations across steps
history = run.scan_history()
history_df = pd.DataFrame(history)

# Find all neuron activation columns
neuron_cols = [col for col in history_df.columns if re.match(r"^neuron_activations/layer_(\d+)/mac_(\d+)/cm_(\d+)/neuron_(\d+)$", col)]

# Group by layer, MAC, and CM
neuron_stats = {}
for col in neuron_cols:
    match = re.match(r"^neuron_activations/layer_(\d+)/mac_(\d+)/cm_(\d+)/neuron_(\d+)$", col)
    if match:
        layer, mac, cm, neuron = match.groups()
        key = f"Layer {layer}, MAC {mac}, CM {cm}"
        if key not in neuron_stats:
            neuron_stats[key] = {}

        # Calculate activation percentage over all steps
        activations = history_df[col].dropna()
        total = len(activations)
        active = sum(activations == 1)
        pct = active / total * 100 if total > 0 else 0
        neuron_stats[key][f"Neuron {neuron}"] = f"{active}/{total} ({pct:.1f}%)"

# Display activation statistics
print("=== Neuron Activation Statistics ===")
for group, neurons in sorted(neuron_stats.items()):
    print(f"\n{group}:")
    for neuron, stats in sorted(neurons.items()):
        print(f"  {neuron}: {stats}")
```

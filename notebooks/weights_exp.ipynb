```python
import wandb
import torch
import os

# Initialize a wandb run
# run = wandb.init(project="weights_testing", entity="sparsey-testing-system", job_type="notebook")

# Use the artifact
# artifact = run.use_artifact('sparsey-testing-system/weights_testing/qut03o6w-weights:v49', type='weights_structure')
api = wandb.Api()

artifact = api.artifact(name='sparsey-testing-system/weights_testing/qut03o6w-weights:v49', type='weights_structure')

# Download the artifact
artifact_dir = artifact.download()
print(f"Artifact downloaded to: {artifact_dir}")

# List files in the artifact directory
print("Files in the artifact directory:")
for file in os.listdir(artifact_dir):
    print(f"- {file}")

# weights

# Find and load .pt files
pt_files = [f for f in os.listdir(artifact_dir) if f.endswith('.pt')]
if pt_files:
    for pt_file in pt_files:
        file_path = os.path.join(artifact_dir, pt_file)
        print(f"\nLoading {pt_file}...")
        
        # Load the PyTorch model or tensor
        weights = torch.load(file_path)
        
        # Display information about the loaded content
        if isinstance(weights, dict):
            print(f"Loaded dictionary with keys: {list(weights.keys())}")
            for key, value in weights.items():
                if isinstance(value, torch.Tensor):
                    print(f"  {key}: Tensor of shape {value.shape}, dtype {value.dtype}")
                else:
                    print(f"  {key}: {type(value)}")
        elif isinstance(weights, torch.Tensor):
            print(f"Loaded tensor of shape {weights.shape}, dtype {weights.dtype}")
            print(f"Tensor statistics: min={weights.min().item()}, max={weights.max().item()}, mean={weights.mean().item()}")
        else:
            print(f"Loaded object of type: {type(weights)}")
else:
    print("No .pt files found in the artifact directory.")
```

```python
import wandb
import torch
import os
import pprint

api = wandb.Api()

run = api.run(
    path='sparsey-testing-system/weights_testing/81di3zhg'
)
run_id = run.id
fetched_weighted_connections = {}

# Ensure the target download directory exists or choose a temporary one
download_dir_base = "/Users/dhruv/indexCode/sparseypy-svs/ruf/artifacts/weights_download"
os.makedirs(download_dir_base, exist_ok=True)

for i, logged_artifact in enumerate(run.logged_artifacts()):
  if "-model" in logged_artifact.name:
    continue
  try:
    artifact = api.artifact(
      f'sparsey-testing-system/weights_testing/{run_id}-weights:v{i}',
      type='weighted_connections'
    )
    print(f'{i}: Downloading {artifact.name}')
    # Download to a unique subdirectory for each artifact version
    # download_dir_base = os.path.join(download_dir_base, f"v{i}")
    # Ensure the specific version directory exists before downloading
    os.makedirs(download_dir_base, exist_ok=True)
    # Use download with root specified to place it in the versioned folder
    artifact.download(root=download_dir_base)

    # Find the .pt file in the downloaded directory
    pt_file = None
    for filename in os.listdir(download_dir_base):
      if filename.endswith('.pt'):
        pt_file = os.path.join(download_dir_base, filename)
        break

    if pt_file:
      print(f'  Loading {pt_file}')
      weights_data = torch.load(pt_file)
      fetched_weighted_connections[i] = weights_data
      print(f'  Stored data for index {i}')
    else:
      print(f'  Warning: No .pt file found in artifact {artifact.name} at {download_dir_base}')

  except Exception as e:
      print(f"Error processing artifact v{i} ({logged_artifact.name}): {e}")

print("\nFinished processing artifacts.")
print(f"Collected data for indices: {list(fetched_weighted_connections.keys())}")

# Optionally print the structure of the fetched data for the first index if it exists
if 0 in fetched_weighted_connections:
    print("\nData structure for index 0:")
    if isinstance(fetched_weighted_connections[0], dict):
        for key, value in fetched_weighted_connections[0].items():
             if isinstance(value, torch.Tensor):
                 print(f"  {key}: Tensor shape {value.shape}")
             else:
                 print(f"  {key}: type {type(value)}")
    elif isinstance(fetched_weighted_connections[0], torch.Tensor):
         print(f"  Tensor shape {fetched_weighted_connections[0].shape}")
    else:
        pprint.pprint(fetched_weighted_connections[0])

# You can now access the data like: fetched_weighted_connections[0], fetched_weighted_connections[1], etc.
```

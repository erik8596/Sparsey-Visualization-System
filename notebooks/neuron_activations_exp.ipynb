```python
import wandb
import torch
import pprint

# init
api = wandb.Api()

entity = 'sparsey-testing-system'
project = 'svs_spring_demo'
run_id = 'jizlsez2'

run = api.run(path=f"{entity}/{project}/{run_id}")
```

```python
res = {}
for i, step in enumerate(run.scan_history()):
    res[i] = step

pprint.pprint(res, sort_dicts=True)
```

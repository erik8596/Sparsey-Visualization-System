```python
import wandb

api = wandb.Api()

project_name = "weights_testing"
entity = "sparsey-testing-system"
project = api.project(
    name=project_name,
    entity=entity
)
runs = api.runs(
    path=f"{entity}/{project_name}"
)
run_id = [run.id for run in runs if "-eval" not in run.name].pop()
print(run_id)
```

```python
import pprint

run = api.run(f"{entity}/{project_name}/{run_id}")
scan = run.scan_history()

for step in scan:
    pprint.pprint(step)
```

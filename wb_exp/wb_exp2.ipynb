```python
import wandb

api = wandb.Api()

entity, project = "sparsey-testing-system", "post_presentation_4"

runs = api.runs(f"{entity}/{project}")

for run in runs:
    print(f"Run ID: {run.id}")
    print(f"Run Name: {run.name}")
    print(f"Run URL: {run.url}")
    print()
```

```python
# weights and biases exploring

# import wandb library and load the api
# wandb should already be logged in because of the virtual environment that's activated both the python and conda environments.

import wandb

api = wandb.Api()

entity, project = "sparsey-testing-system", "visualization-data-fetching"

runs = api.runs(f"{entity}/{project}")

for run in runs:
    print(f"Run ID: {run.id}")
    print(f"Run Name: {run.name}")
    print(f"Run URL: {run.url}")
    print()
```

```python
# pull experiment config data and a summary of one of the training runs

run_id = "tex3jfof"
run = api.run(f"{entity}/{project}/{run_id}")

print(f"Run ID: {run.id} | Run Name: {run.name}\n")

# print("Config:")
# for key, value in run.config.items():
#     if isinstance(value, dict):
#         print(f"{key}:")
#         for k, v in value.items():
#             print(f"{k}: {v}")
#     else:
#         print(f"{key}: {value}")

print()

print("Summary:")
for key, value in run.summary.items():
    print(f"{key}: {value}")
```

```python
# history of a specific run

history = run.history(
    samples=500, keys=None, x_axis="_step", pandas=(True), stream="default"
)

history.head(len(history))
```

```python
# scan_history function

scan_history = run.scan_history(
    # keys=None, page_size=1000, min_step=None, max_step=None
)

for row in scan_history:
    for key, value in row.items():
        print(f"{key}: {value}")
    print()
    print()
```

```python
# evaluation run

run_id_eval = "cf4wgk9a"
run_eval = api.run(f"{entity}/{project}/{run_id_eval}")

print(f"Run ID: {run_eval.id} | Run Name: {run_eval.name}\n")

print("Config:")
for key, value in run_eval.config.items():
    if isinstance(value, dict):
        print(f"{key}:")
        for k, v in value.items():
            print(f"{k}: {v}")
    else:
        print(f"{key}: {value}")

print()

print("Summary:")
for key, value in run_eval.summary.items():
    print(f"{key}: {value}")
```
